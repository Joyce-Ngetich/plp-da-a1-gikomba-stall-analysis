```python
# Provided Dataset
daily_sales = [
    2300,
    1950,
    2800,
    2100,
    3600,
    5200,
    4100,
    2500,
    2200,
    3100,
    2400,
    3900,
    5600,
    4700,
]  # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Basic Sales Computations
**Approach Explanation:** Calculated total, average, highest, and lowest sales across the 14-day list using built-in functions. Used `.index()` to locate best and worst performing days and mapped their positions to specific weekdays and week numbers.

```python
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)

max_sale = max(daily_sales)
min_sale = min(daily_sales)

max_index = daily_sales.index(max_sale)
min_index = daily_sales.index(min_sale)

best_day_name = days[max_index]
best_week = (max_index // 7) + 1

worst_day_name = days[min_index]
worst_week = (min_index // 7) + 1

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(
    f"Highest Sale: KES {max_sale:,} on Week {best_week} ({best_day_name})"
)
print(f"Lowest Sale: KES {min_sale:,} on Week {worst_week} ({worst_day_name})")
```

## Task 2: Week-on-Week Growth Analysis
**ApproachExplanation:** Created a modular function `week_totals(sales)` utilizing list slicing to separate week 1 and week 2 data, then computed the percentage growth between the two periods.

```python
def week_totals(sales):
  """Takes a 14-day sales list, slices into two 7-day periods,

  and returns a tuple of (week_1_total, week_2_total).
  """
  w1_total = sum(sales[:7])
  w2_total = sum(sales[7:])
  return w1_total, w2_total


w1, w2 = week_totals(daily_sales)
wow_growth = ((w2 - w1) / w1) * 100

print(f"Week 1 Total: KES {w1:,}")
print(f"Week 2 Total: KES {w2:,}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

## Task 3: Day Classification
Approach Explanation: Defined `classify_day()` to evaluate daily performance against $\pm 20\%$ thresholds of the mean. Applied it across all 14 days and aggregated counts into a category dictionary.

```python
def classify_day(amount, average):
  """Classifies a daily sales amount as 'Strong', 'Normal', or 'Weak'

  based on whether it deviates more than 20% above or below the average.
  """
  if amount > average * 1.2:
    return "Strong"
  elif amount < average * 0.8:
    return "Weak"
  else:
    return "Normal"


category_counts = {}
for sale in daily_sales:
  cat = classify_day(sale, avg_sales)
  category_counts[cat] = category_counts.get(cat, 0) + 1

print("Day Classifications Count:")
for category, count in category_counts.items():
  print(f"- {category}: {count} days")
```

## Task 4: Weekdays vs Weekends Comparison
**Approach Explanation:** Filtered sales by weekday and weekend indices across both weeks to compute group averages and evaluate the percentage premium of weekend shopping.

```python
# Weekdays indices: Mon-Fri of week 1 (0:5) and week 2 (7:12)
weekday_sales = daily_sales[0:5] + daily_sales[7:12]
# Weekends indices: Sat-Sun of week 1 (5:7) and week 2 (12:14)
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
weekend_premium = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend Sales Premium: {weekend_premium:.2f}% higher than weekdays")
```

## Task 5: Inventory Reorder & Gross Margins
**Approach Explanation:** Iterated through the stock dictionary using `.items()` to identify items at or below reorder levels and calculated individual product gross margin percentages.

```python
def check_reorders_and_margins(stock_dict):
  """Identifies products needing restock and computes gross margin percentage

  for each item in the stock dictionary.
  """
  reorder_list = []
  margins = {}

  for product, info in stock_dict.items():
    if info["qty"] <= info["reorder"]:
      reorder_list.append(product)

    margin = ((info["price"] - info["cost"]) / info["price"]) * 100
    margins[product] = margin

  return reorder_list, margins


items_to_restock, product_margins = check_reorders_and_margins(stock)

print("Items requiring reorder:", items_to_restock)
print("\nGross Margins by Product:")
for prod, margin in product_margins.items():
  print(f"- {prod}: {margin:.2f}%")
```

## Task 6: Stall Performance Report
**Approach Explanation:** Generated a professionally formatted executive summary using f-strings with thousands separators for monetary figures.

```python
print("=" * 50)
print("          STALL PERFORMANCE REPORT          ")
print("=" * 50)
print(f"Total 14-Day Revenue    : KES {total_sales:,}")
print(f"Week 1 Revenue          : KES {w1:,}")
print(f"Week 2 Revenue          : KES {w2:,}")
print(f"Week-on-Week Growth     : {wow_growth:.2f}%")
print(f"Average Daily Sales     : KES {avg_sales:,.2f}")
print(f"Average Weekday Sales   : KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales   : KES {avg_weekend:,.2f}")
print(f"Best Sales Day          : KES {max_sale:,} ({best_day_name}, W{best_week})")
print(f"Lowest Sales Day        : KES {min_sale:,} ({worst_day_name}, W{worst_week})")
print("-" * 50)
print("INVENTORY STATUS & MARGINS:")
for prod, info in stock.items():
  status = "RESTOCK NEEDED" if prod in items_to_restock else "Stock OK"
  margin = product_margins[prod]
  print(
      f"• {prod:<10} | Qty: {info['qty']:2d} kg | Reorder: {info['reorder']:2d} kg"
      f" | Margin: {margin:5.1f}% | Status: {status}"
  )
print("=" * 50)
```
